```python
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python

def process_pdf_folder(pdf_directory, chunk_size=1000, chunk_overlap=200):
    """
    Iterates through a folder of PDF files, loads them, chunks the content,
    and returns a list of dictionaries with Product name and Description.
    """
    structured_data = []
    
    # Initialize the text splitter for chunking
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len
    )
    
    # Check if directory exists
    if not os.path.exists(pdf_directory):
        print(f"Directory '{pdf_directory}' does not exist.")
        return structured_data

    # Iterate through all files in the directory one by one
    for filename in sorted(os.listdir(pdf_directory)):
        if filename.lower().endswith(".pdf"):
            file_path = os.path.join(pdf_directory, filename)
            
            # Extract product name from the filename (removing the .pdf extension)
            product_name = os.path.splitext(filename)[0]
            
            print(f"Processing: {filename}")
            
            try:
                # Load the PDF using PyPDFLoader
                loader = PyPDFLoader(file_path)
                pages = loader.load()
                
                # Split pages into chunks
                chunks = text_splitter.split_documents(pages)
                
                # Create a record for each chunk or combine text if preferred
                # Here we create a dictionary for each chunk's content
                for chunk in chunks:
                    structured_data.append({
                        "Product name": product_name,
                        "Description": chunk.page_content
                    })
                    
            except Exception as e:
                print(f"Error processing {filename}: {e}")
                
    return structured_data

```

```python
# --- Example Usage ---
if __name__ == "__main__":
    # Replace with the path to your folder containing the 509 PDFs
    PDF_FOLDER_PATH = "./path_to_your_pdfs"
    
    # Run the processing pipeline
    product_chunks_list = process_pdf_folder(PDF_FOLDER_PATH)
    
    print(f"\nSuccessfully processed and created {len(product_chunks_list)} total chunks.")
    
    # Print the first item as a sample if available
    if product_chunks_list:
        print("\nSample Output:")
        print(product_chunks_list[0])
```
